### 3.3. Простой линейный график

Для начала попробуем создать самый простой график, показывающий, как сумма `amount` меняется по месяцам. Для этого используется метод `plt.plot()` нашей библиотеки:

````python
# запустите ячейку кода

import matplotlib.pyplot as plt

months = ["Янв", "Фев", "Мар", "Апр", "Май"]
amounts = [50000, 62000, 58000, 71000, 68000]

plt.plot(months, amounts)
plt.show()
````

### 3.4. Настройка графика
В Matplotlib есть масса возможностей для проработки визуала графика. 

**Основные параметры:**

| Параметр | Что делает |
|----------|------------|
| `figsize=(w, h)` | Размер полотна в дюймах |
| `color` | Цвет линии |
| `marker` | Маркер точек (`o`, `s`, `^`, `*`) |
| `linestyle` | Стиль линии (`-`, `--`, `:`, `-.`) |
| `alpha` | Прозрачность (0–1) |
| `linewidth` | Толщина линии | 

Название горизонтальной оси графика 
**Настройка областей графика:**

| Параметр | Что делает |
|----------|------------|
| `plt.xlabel` | Размер полотна в дюймах |
| `plt.ylabel` | Цвет линии |
| `plt.title` | Маркер точек (`o`, `s`, `^`, `*`) |
| `linestyle` | Стиль линии (`-`, `--`, `:`, `-.`) |
| `alpha` | Прозрачность (0–1) |
| `linewidth` | Толщина линии | 



```python
plt.figure(figsize=(10, 6))         # размер полотна
plt.plot(months, amounts, color="steelblue", marker="o")
plt.title("Динамика транзакций по месяцам")
plt.xlabel("Месяц")
plt.ylabel("Сумма, руб.")
plt.grid(True, alpha=0.3)
plt.show()
```



### 3.5. Гистограмма

**Гистограмма** показывает распределение значений: сколько значений попадает в каждый интервал.

```python
import numpy as np

amounts = np.random.normal(50000, 15000, 200)

plt.figure(figsize=(10, 6))
plt.hist(amounts, bins=20, color="steelblue", edgecolor="black")
plt.title("Распределение сумм транзакций")
plt.xlabel("Сумма, руб.")
plt.ylabel("Количество транзакций")
plt.show()
```

**Параметр `bins`** — количество столбцов. Чем больше данных, тем больше столбцов.

### 3.6. Boxplot (ящик с усами)

**Boxplot** показывает:
- медиану (центральная линия);
- Q1 и Q3 (границы ящика);
- границы выбросов (усы);
- сами выбросы (точки за усами).

```python
plt.figure(figsize=(8, 6))
plt.boxplot(amounts)
plt.title("Boxplot сумм транзакций")
plt.ylabel("Сумма, руб.")
plt.show()
```

**Boxplot — лучший график для поиска выбросов.**

### 3.7. Столбчатая диаграмма

```python
contractors = ["Контрагент_A", "Контрагент_B", "Контрагент_C", "Контрагент_D"]
totals = [250000, 180000, 320000, 95000]

plt.figure(figsize=(10, 6))
plt.bar(contractors, totals, color="coral", edgecolor="black")
plt.title("Сумма транзакций по контрагентам")
plt.xlabel("Контрагент")
plt.ylabel("Сумма, руб.")
plt.xticks(rotation=45)
plt.show()
```

### 3.8. Диаграмма рассеяния (scatter)

```python
x = [100000, 150000, 200000, 250000, 300000]
y = [20000, 30000, 40000, 48000, 60000]

plt.figure(figsize=(8, 6))
plt.scatter(x, y, color="green", alpha=0.7)
plt.title("Связь выручки и налогов")
plt.xlabel("Выручка, руб.")
plt.ylabel("Налоги, руб.")
plt.grid(True, alpha=0.3)
plt.show()
```

### 3.9. Несколько графиков на одном полотне

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)     # 1 строка, 2 столбца, ячейка 1
plt.plot(months, amounts)
plt.title("Динамика")

plt.subplot(1, 2, 2)     # ячейка 2
plt.hist(amounts, bins=10)
plt.title("Распределение")

plt.tight_layout()
plt.show()
```

### 3.10. Сохранение графика

```python
plt.figure(figsize=(10, 6))
plt.plot(months, amounts)
plt.title("Динамика")
plt.savefig("chart.png", dpi=150)
plt.show()
```

Параметр `dpi` задаёт качество картинки.

### 3.12. Гистограмма + плотность распределения

```python
sns.histplot(data=df, x="amount", bins=20, kde=True)
```

**`kde=True`** — сглаженный график плотности поверх гистограммы.

### 3.13. Boxplot в seaborn

```python
sns.boxplot(data=df, x="contractor", y="amount")
```

**С группировкой по цвету:**

```python
sns.boxplot(data=df, x="contractor", y="amount", hue="risk")
```

### 3.14. Столбчатая диаграмма в seaborn

```python
sns.barplot(data=df, x="contractor", y="amount", estimator="sum")
```

По умолчанию `barplot` показывает среднее. Аргумент `estimator="sum"` показывает сумму.

### 3.15. Диаграмма рассеяния в seaborn

```python
sns.scatterplot(data=df, x="amount", y="tax", hue="risk", size="amount")
```

- `hue` — цвет точек по категории;
- `size` — размер точек по числовому значению.

### 3.16. Графики по группам (FacetGrid)

```python
g = sns.FacetGrid(df, col="risk", col_wrap=2, height=4)
g.map(plt.hist, "amount", bins=10)
```

Это создаёт отдельный график для каждой категории риска.

### 3.17. Heatmap корреляций

```python
corr = df[["amount", "tax", "staff", "age"]].corr()
sns.heatmap(corr, annot=True, cmap="coolwarm", vmin=-1, vmax=1)
```

Показывает корреляции между числовыми столбцами.

## 7. Типичные ошибки и как их избежать

| Ошибка | Причина | Решение |
|--------|---------|---------|
| График не отображается | Нет `%matplotlib inline` или `plt.show()` | Добавить `plt.show()` |
| Графики накладываются друг на друга | Не вызван `plt.figure()` перед каждым | Разделить через `plt.figure()` |
| Подписи перекрываются | Мало места | Использовать `plt.tight_layout()` |
| Русские буквы не отображаются | Проблема с шрифтом | Использовать `plt.rcParams` или писать латиницей |
| `ValueError: x and y must have same first dimension` | Разная длина списков | Проверить длины |
| График слишком маленький | Не задан `figsize` | Добавить `plt.figure(figsize=(10, 6))` |
| Файл не сохраняется | Неверный путь | Проверить рабочую папку |
| `KeyError` в seaborn | Опечатка в названии столбца | Проверить `df.columns` |
| Гистограмма некрасивая | Мало/много `bins` | Подобрать `bins` (10–30) |
| Слишком много цветов в легенде | Много категорий | Ограничить или использовать палитры |
