## Приложение. Пример данных

Файл `control.csv` может быть сгенерирован так:

```python
import pandas as pd
import random

random.seed(42)

contractors = [
    "Контрагент_A", "Контрагент_B", "Контрагент_C",
    "Контрагент_D", "Контрагент_E"
]
risks = ["низкий", "средний", "высокий", "критический"]
types = ["поступление", "списание"]

rows = []
for i in range(1, 31):
    contractor = random.choice(contractors)
    amount = random.randint(3000, 30000)
    if i == 15:
        amount = 500000   # выброс
    if i == 27:
        amount = 400000   # выброс
    risk = random.choice(risks)
    op_type = random.choice(types)
    date = f"2024-{random.randint(1, 12):02d}-{random.randint(1, 28):02d}"
    rows.append({
        "id": i,
        "date": date,
        "contractor": contractor,
        "amount": amount,
        "risk": risk,
        "type": op_type
    })

df = pd.DataFrame(rows)
df.to_csv("control.csv", index=False)
print("Файл control.csv создан")
```
