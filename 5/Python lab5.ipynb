### 3.2. NumPy: массивы

NumPy-массив — это структура данных, похожая на список, но:
- все элементы одного типа;
- операции выполняются **поэлементно** (векторизованно);
- массив занимает меньше памяти.

**Векторизация** предполагает представление всего массива данных как структурной единицы - вектора. И работа идет со всей единицей сразу, а не с её отдельными элементами. По этой причине так называемые векторные операции не требуют от нас обработки массива через цикл `for`.

### 3.3. NumPy: типы данных

Все элементы массива имеют один тип. Для определения типа данных служит метод `.dtype`:

````python
arr_int = np.array([10, 20, 30])
print(arr_int.dtype)   # → int64, массив целых чисел

arr_float = np.array([10.5, 20.3, 30.1])
print(arr_float.dtype)  # → float64, массив чисел с дробной частью

arr_bool = np.array([True, False, True])
print(arr_bool.dtype)   # → bool, массив логических значений
````

**Преобразование типа выполняется через `astype`:**

````python
arr = np.array([10.7, 20.3, 30.9])
arr_int = arr.astype('int') # метод astype() изменяет исходный тип на данных на указанный в скобках
print(arr_int)   # → [10 20 30]
````

### 3.4. NumPy: основные функции

| Функция              | Что делает                                                             |
|----------------------|------------------------------------------------------------------------|
| `arr.mean()`         | Среднее арифметическое всех элементов массива arr                          |
| `arr.std()`          | Стандартное отклонение элементов массива arr                              |
| `np.median(arr)`     | Медиана массива arr (центральное значение отсортированного массива)        |
| `arr.min()`          | Минимальный элемент массива arr                                           |
| `arr.max()`          | Максимальный элемент массива arr                                           |
| `arr.sum()`          | Сумма всех элементов массива arr                                           |
| `arr.size`           | Количество элементов в массиве arr                                         |

````python
arr = np.array([10, 20, 30, 40, 50])

print(arr.mean())     # → 30.0
print(arr.std())      # → 14.142... (стандартное отклонение)
print(np.median(arr)) # → 30.0 
print(arr.min())      # → 10
print(arr.max())      # → 50
print(arr.sum())      # → 150
print(arr.size)       # → 5 (количество элементов)
````

### 3.5. NumPy: фильтрация элементов

Для отсеивания элементов массива используются так называемые *булевы маски*. Можно применить оператор сравнения ко всему массиву целиком, и результатом такой операции будет аналогичный по размеру массив, состоящий из `True` (если для данного элемента условие верно) и `False` (если неверно)

````python
arr = np.array([10, 20, 30, 40, 50])
print(arr > 25)               # → [False False  True  True  True]
````

для первых двух элементов массива (10 и 20) условие `>25` неверно, поэтому в ячейках стоит `False`. Для оставшихся трех условие выполняется, поэтому записаны значения `True`.
Созданную маску можно накладывать на исходный массив, помещая в квадратные скобки `[]`:

````python
print(arr[arr > 25])          # → [30 40 50]
````

после применения маски в новый массив `arr[arr > 25]` попадут только те элементы, для которых условие маски имеет значение `True` (30, 40, 50).

Булева маска может иметь сложную структуру из нескольких условий.

**Важно:** для объединения условий использовать `&` и `|`, а не `and`/`or`. Каждое условие — в скобках.

````python
# выведем на печать элементы, которые одновременно больше 15 и меньше 45
print(arr[(arr > 15) & (arr < 45)])   # → [20 30 40]
````

### 3.13. Применение функции через `apply`
Метод позволяет применить любую функцию к выбранному столбцу датафрейма. В примере создана функция категоризации сделок по сумме `categorize()`, далее она применяется к столбцу `df["amount"]`, а результат записывается в новый столбец `df["category"]`.

````python
def categorize(amount): # создание функции через def
    if amount > 50000:
        return "крупная"
    elif amount > 10000:
        return "средняя"
    else:
        return "мелкая"

df["category"] = df["amount"].apply(categorize) #применяет функцию categorize() к столбцу amount
````

Аналогично метод работает **с `lambda` - функциями:**

````python
df["amount_doubled"] = df["amount"].apply(lambda x: x * 2)
````

### 3.14. Группировка через `groupby`
Метод позволяет разделить данные на отдельные группы по критерию (например, значению определенного столбца). Далее в каждой группе проводятся вычисления: считается среднее, находится количество элементов, их сумма, минимум/максимум и т.д.

**Пример 1**
Сгруппируем наш датафрейм `df` по столбцу `risk`, и пересчитаем, сколько контрагентов датасета входит в каждую группу риска (за пересчет отвечает метод `.count()`):

````python
df.groupby("risk")["contractor"].count()
````

**Пример 2**
Сгруппируем по столбцу `risk`, и найдем среднюю сумму сделки в каждой группе риска (за нахождение среднего значения отвечает метод `.mean()`):

````python
df.groupby("risk")["amount"].mean()
````

Возможно рассчитать **несколько показателей** на одной и той же группировке, используя метод `.agg(список_нужных_показателей)`.
**Найдем** количество контрагентов, сумму всех сделок, среднюю сумму сделки, медиану и максимум в каждой группе риска:

````python
df.groupby("risk")["amount"].agg(["count", "sum", "mean", "median", "max"])
````
