## 3.8. Условные операторы


```python
if условие:
    # действия, если условие истинно
elif другое_условие:
    # действия, если первое ложно, а второе истинно
else:
    # действия, если все условия ложны
```
### Пример

````python
score = 75
if score >= 80:
    print("Высокий риск")
elif score >= 50:
    print("Средний риск")
else:
    print("Низкий риск")
````

### Логические операторы

| Оператор | Значение | Пример |
|----------|----------|--------|
| `and` | И (оба истинны) | `x > 0 and x < 10` |
| `or` | Или (хотя бы один истинен) | `x < 0 or x > 10` |
| `not` | Не (отрицание) | `not (x > 0)` |

````python
amount = 1500000
staff = 1
if amount > 1000000 and staff == 1:
    print("Признак фирмы-однодневки")
else:
    print("Признаков не выявлено")
````

### Операторы сравнения

| Оператор | Значение |
|----------|----------|
| `==` | Равно |
| `!=` | Не равно |
| `>` | Больше |
| `<` | Меньше |
| `>=` | Больше или равно |
| `<=` | Меньше или равно |

### Многоуровневое условие

## 3.9. Цикл `while`

```python
while условие:
    # действия, пока условие истинно
```

### Пример — простой счётчик

````python
count = 0
while count < 5:
    print(f"Итерация {count}")
    count = count + 1
````


---

### Пример — накопление суммы


````python
total = 0
count = 0
while count < 3:
    amount = float(input("Введите сумму транзакции: "))
    total = total + amount
    count = count + 1
print(f"Итого: {total:,.2f} руб.")
````

### Пример — поиск первой крупной транзакции с `break`

`break` используется для принудительного прерывания цикла по достижении результата.

````python
amounts = [50000, 120000, 300000, 80000]
i = 0
while i < len(amounts):
    if amounts[i] > 100000:
        print(f"Первая крупная транзакция: {amounts[i]} руб. (индекс {i})")
        break
    i = i + 1
````

**Важно:** цикл `while` может выполняться бесконечно, если условие всегда истинно. Следите за тем, чтобы условие рано или поздно стало ложным.

### Сокращённые операторы

Вместо `count = count + 1` можно писать `count += 1`:

````python
count = 0
while count < 5:
    print(count)
    count += 1
````

Аналогично: `total += amount`, `x -= 1`, `y *= 2`, `z /= 3`.
