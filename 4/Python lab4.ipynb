## 3. Теоретический материал

### 3.1. Зачем нужны функции

В лабораторных №1–3 мы писали код для конкретной задачи: скоринговая модель, описательная статистика, скрининг. Но в реальной работе одни и те же вычисления приходится повторять: сегодня — для одного контрагента, завтра — для другого, послезавтра — для 50 контрагентов.

> **Функция** — это именованный блок кода, который можно вызывать многократно с разными данными. Функция — это «инструмент» в вашей библиотеке.

Например, чтобы найти среднее, мы каждый раз писали:

```python
mean = sum(amounts) / len(amounts)
```

А можно один раз создать функцию:

```python
def calc_mean(data):
    return sum(data) / len(data)
```

И вызывать её:

```python
calc_mean([5000, 12000, 8000])
calc_mean([4500, 6100, 7200, 8900])
```

### 3.2. Объявление функции

```python
def имя_функции(параметры):
    # тело функции
    return результат
```

**Простая функция:**

```python
def greet():
    print("Привет, студент!")
```

**Вызов:**

```python
greet()   # → Привет, студент!
```

### 3.3. Параметры и аргументы

**Параметр** — это имя переменной в объявлении функции. **Аргумент** — это значение, которое передаётся при вызове.

```python
def square(x):  # x  - это параметр
    return x ** 2

print(square(5))    # 5 - это аргумент, функция вернет значение → 25
print(square(10))   # 10  - это аргумент, функция вернет значение → 100
```

**Несколько параметров:**

```python
def total_with_tax(amount, tax_rate):
    return amount * (1 + tax_rate)

print(total_with_tax(1000, 0.2))   # → 1200.0
```

### 3.4. Возврат значения `return`

`return` возвращает результат из функции и завершает её выполнение.

```python
def calc_mean(data):
    return sum(data) / len(data)

result = calc_mean([5000, 12000, 8000])
print(result)   # → 8333.333333333334
```

**Если `return` отсутствует** — функция возвращает `None`:

```python
def say_hello():
    print("Hello")

x = say_hello()    # → Hello
print(x)           # → None
```

**Возврат нескольких значений (кортеж):**

```python
def stats(data):
    return min(data), max(data), sum(data) / len(data)  # возвращаем сразу три результата

minimum, maximum, mean = stats([5000, 12000, 8000]) # получаем все три результата и записываем их в переменные minimum, maximum, mean соответственно 
print(minimum, maximum, mean)
```

### 3.5. Значения по умолчанию

```python
def total_with_tax(amount, tax_rate=0.2):  # amountm tax_rate - это параметры функции, при этом tax_rate имеет значение по умолчанию 0.2
    return amount * (1 + tax_rate)

print(total_with_tax(1000))          # если не указать явно tax_rate в аргументах, функция использует значение по умолчанию и вернет → 1200.0
print(total_with_tax(1000, 0.1))     # можно указать tax_rate=0/1, и функция будет использовать новое значение и вернет → 1100.0
```

**Важно:** параметры со значениями по умолчанию должны идти **после** параметров без значений.

### 3.6. Области видимости

**Локальная переменная** — создана внутри функции, доступна только внутри неё.

```python
def calc_mean(data):
    total = sum(data)          # локальная
    return total / len(data)

calc_mean([5000, 12000])
print(total)   # → NameError: total не определена
```

**Глобальная переменная** — создана вне функции в основной программе, доступна везде, но изменять её внутри функции не рекомендуется.

```python
TAX_RATE = 0.2   # глобальная константа

def total_with_tax(amount):
    return amount * (1 + TAX_RATE)
```

**Правило:** функция должна зависеть только от своих параметров. Не используйте глобальные переменные внутри функции — это плохая практика.

### 3.7. Docstring — документация функции

```python
def calc_mean(data):
    """
    Вычисляет среднее арифметическое списка чисел.
    
    Параметры:
        data (list): список чисел
    
    Возвращает:
        float: среднее арифметическое
    """
    return sum(data) / len(data)
```

Docstring — это многострочный комментарий в первой строке функции. Он помогает другим (и вам через месяц) понять, что делает функция.

### 3.8. `lambda`-функции

**`lambda`** — это короткая анонимная функция. Используется для простых операций, когда объявлять полноценную функцию через `def` избыточно.

```python
square = lambda x: x ** 2
print(square(5))   # → 25
```

**Синтаксис:**

```python
lambda параметры: выражение
```

**Примеры:**

```python
add = lambda a, b: a + b
is_large = lambda x: x > 100000
to_percent = lambda x: x / 100
```

### 3.9. `map` и `filter`

**`map(функция, список)`** — применяет функцию к каждому элементу списка.

```python
amounts = [5000, 12000, 8000]
doubled = list(map(lambda x: x * 2, amounts)) #применяет лямбда-функцию умножения на 2 к каждому элементу списка amounts
print(doubled)   # → [10000, 24000, 16000]
```

**`filter(функция, список)`** — оставляет только те элементы, для которых функция вернула `True`.

```python
amounts = [5000, 12000, 8000, 15000]
large = list(filter(lambda x: x > 10000, amounts))
print(large)   # → [12000, 15000]
```

**Важно:** `map` и `filter` возвращают специальные объекты. Чтобы получить список, оберните их в `list()`.

**Сравнение с циклом:**

```python
# Через цикл
doubled = []
for a in amounts:
    doubled.append(a * 2)

# Через map
doubled = list(map(lambda x: x * 2, amounts))

# Через списковое включение (list comprehension)
doubled = [a * 2 for a in amounts]
```

Три способа — один результат. Списковые включения обычно удобнее.

### 3.10. Коэффициент вариации

**Коэффициент вариации ($C_V$)** — это относительная мера разброса. Показывает, какую долю от среднего составляет стандартное отклонение.

**Формула:**

$$C_V = \frac{\sigma}{\bar{x}} \cdot 100\%$$

**Интерпретация:**

| Значение $C_V$ | Однородность совокупности |
|-------------|---------------------------|
| < 33 % | Однородная |
| 33 – 67 % | Средняя неоднородность |
| > 67 % | Неоднородная |

**Реализация:**

```python
import math

def coefficient_of_variation(data):
    n = len(data)
    mean = sum(data) / n
    diffs = sum((x - mean) ** 2 for x in data)
    sigma = math.sqrt(diffs / (n - 1))
    return sigma / mean * 100
```

### 3.11. Индекс концентрации ($C_R$)

**Индекс концентрации $C_{R-N}$** — это доля топ-N значений в общей сумме. Помогает выявить, насколько выручка, платежи или оборот зависят от нескольких крупных клиентов.

**Формула для C_{R-3}:**

$$C_{R-3} = \frac{\text{сумма топ-3 значений}}{\text{сумма всех значений}} \cdot 100\%$$

**Реализация:**

```python
def concentration_ratio(data, n=3):
    sorted_data = sorted(data, reverse=True)
    top_n_sum = sum(sorted_data[:n])
    return top_n_sum / sum(data) * 100
```

**Интерпретация:**
- $C_{R-3}$ < 50 % — низкая концентрация, всё хорошо;
- $C_{R-3}$ от 50 до 80 % — средняя концентрация;
- $C_{R-3}$ > 80 % — высокая концентрация, зависимость от нескольких клиентов.

## 7. Типичные ошибки и как их избежать

| Ошибка | Причина | Решение |
|--------|---------|---------|
| `NameError: name 'x' is not defined` | Локальная переменная вне функции | Использовать `return` |
| `TypeError: missing 1 required positional argument` | Забыт аргумент | Проверить вызов |
| `SyntaxError: non-default argument follows default argument` | Неверный порядок параметров | Перенести значения по умолчанию в конец |
| `UnboundLocalError` | Изменение глобальной переменной без `global` | Использовать параметры |
| `map` возвращает объект, а не список | Забыли `list()` | Обернуть в `list()` |
| Функция возвращает `None` | Забыли `return` | Добавить `return` |
| `lambda` не работает с блоком | `lambda` — только выражение | Использовать `def` для сложной логики |
| CV = 0 | Все данные одинаковы | Это нормально; в делении защититься от нуля |
| CR-N > 100 % | Ошибка в расчёте | Проверить суммирование |
