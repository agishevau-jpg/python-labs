### 2.4. Срезы

**Срез** — это часть списка. Записывается как `[начало:конец]`, причём **правый конец не включается**.

```python
print(amounts[1:3])   # → [12000, 8000]
print(amounts[:2])    # → [5000, 12000]
print(amounts[2:])    # → [8000, 15000, 7000]
print(amounts[:])     # → весь список (копия)
print(amounts[2:2])   # → [] (пустой срез)
```

**Важно:** срез `[:]` создаёт **копию** списка, а не ссылку на него.

### 2.7. Копирование списков

**Важно:** `L2 = L1` создаёт **ссылку**, а не копию.

```python
L1 = [1, 2, 3]
L2 = L1
L2[0] = 100
print(L1)   # → [100, 2, 3] — L1 тоже изменился, хотя мы изменили только L2!
```

**Правильное копирование списка:**

```python
L1 = [1, 2, 3]
L2 = L1.copy()   # или L1[:]
L2[0] = 100
print(L1)        # → [1, 2, 3] — не изменился
```

### 2.9. Описательная статистика

**Описательная статистика** — это совокупность показателей, которые описывают выборку: как выглядит «типичное» значение и как сильно данные разбросаны.

**Меры центральной тенденции:**

| Показатель | Что показывает |
|------------|----------------|
| Среднее | «Типичное» значение, но не устойчиво к выбросам |
| Медиана | Центральное значение, устойчиво к выбросам |
| Мода | Наиболее часто встречающееся значение |

**Меры разброса:**

| Показатель | Что показывает |
|------------|----------------|
| Размах | Разница между максимумом и минимумом |
| IQR | Разница между 75-м и 25-м процентилями |
| Стандартное отклонение | Среднее расстояние от среднего (в лабораторной №3) |


### 2.10. Среднее арифметическое

**Формула:**

$$\bar{x} = \frac{x_1 + x_2 + \dots + x_n}{n}$$

**Реализация в Python:**

```python
amounts = [5000, 12000, 8000, 15000, 7000]
mean = sum(amounts) / len(amounts)
print(mean)   # → 9400.0
```

### 2.11. Медиана

**Алгоритм:**
1. Отсортировать данные по возрастанию.
2. Если количество элементов **нечётное** — взять средний.
3. Если **чётное** — взять среднее двух средних.

```python
amounts = [5000, 12000, 8000, 15000, 7000]
sorted_amounts = sorted(amounts)   # [5000, 7000, 8000, 12000, 15000]
n = len(sorted_amounts)            # 5
middle = n // 2                    # 2
median = sorted_amounts[middle]    # → 8000
```

**Для чётного количества:**

```python
amounts = [5000, 12000, 8000, 15000]
sorted_amounts = sorted(amounts)   # [5000, 8000, 12000, 15000]
n = len(sorted_amounts)            # 4
left = sorted_amounts[n//2 - 1]    # 8000
right = sorted_amounts[n//2]       # 12000
median = (left + right) / 2        # → 10000.0
```

### 2.12. Мода

**Мода** — значение, которое встречается чаще всего.

```python
amounts = [5000, 12000, 8000, 5000, 7000, 5000]
counts = {}
for a in amounts:
    if a not in counts:
        counts[a] = 1
    else:
        counts[a] += 1
# counts = {5000: 3, 12000: 1, 8000: 1, 7000: 1}
mode = max(counts, key=counts.get)
print(mode)   # → 5000
```

### 2.14. Выбросы и IQR

**Выброс** — значение, которое сильно отличается от остальных.

**25-й процентиль** (первый или нижний квартиль, Q1) — это такое число на шкале измерений, меньше которого или равное которому находится ровно 25% всех значений в наборе данных. Оставшиеся 75% данных при этом лежат выше этого значения.

**50-й процентиль** (второй или средний квартиль, Q2) — это **медиана**. Такое число на шкале измерений, меньше которого или равное которому находится ровно 50% всех значений в наборе данных. Оставшиеся 50% данных при этом лежат выше этого значения

**75-й процентиль** (третий или верхний квартиль, Q3) — это такое число на шкале измерений, меньше которого или равное которому находится ровно 75% всех значений в наборе данных. Оставшиеся 25% данных при этом лежат выше этого значения

**Метод IQR (межквартильный размах):**

1. Найдите Q1 — 25-й процентиль.
2. Найдите Q3 — 75-й процентиль.
3. IQR = Q3 − Q1.
4. Границы: `[Q1 − 1.5·IQR; Q3 + 1.5·IQR]`.
5. Всё, что вне этих границ — выбросы.

```python
amounts = [5000, 12000, 8000, 15000, 7000, 500000]
sorted_amounts = sorted(amounts)

n = len(sorted_amounts)
q1 = sorted_amounts[n // 4]
q3 = sorted_amounts[3 * n // 4]
iqr = q3 - q1

lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr

outliers = [a for a in amounts if a < lower or a > upper]
print(outliers)   # → [500000]
```

### 2.15. Почему медиана устойчивее среднего

**Пример:**

```python
amounts_normal = [5000, 7000, 8000, 12000, 15000]
amounts_outlier = [5000, 7000, 8000, 12000, 500000]

print(sum(amounts_normal)/len(amounts_normal))    # → 9400
print(sum(amounts_outlier)/len(amounts_outlier))  # → 106400 (в 11 раз больше!)
```

При этом медиана в обоих случаях останется около 8000.

> **Вывод:** среднее «ломается» от одного выброса, медиана — нет. Поэтому для анализа зарплат, цен, выручки всегда смотрят на медиану.
